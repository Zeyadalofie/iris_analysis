### Step 6: Plot the Data Using Matplotlib

In this section, we use Matplotlib to visualize the dataset. We create a bar chart to compare the average petal length of each Iris species, making it easier to understand the differences between the species.

```python
import matplotlib.pyplot as plt

mean_Petal_Length = df.groupby('species')['petal_length'].mean()

mean_Petal_Length.plot(kind='bar')

plt.title('Average Petal Length by Species')
plt.xlabel('Species')
plt.ylabel('Average Petal Length (cm)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
